```python
import pandas as pd
df1 = pd.DataFrame({
    'ID': [1, 2, 3],
    'Name': ['Asha', 'Ravi', 'Sita'],
    'Marks': [85, 90, 78]
})
print("DataFrame 1:\n", df1)
```

```python
# 2. Hierarchical Indexing
df = pd.DataFrame({
    'Marks': [85, 90, 78, 88]
}, index=pd.MultiIndex.from_tuples([
    ('CSE', 1), ('CSE', 2), ('ECE', 1), ('ECE', 2)
], names=['Branch', 'Student']))
print("\nHierarchical Indexing:\n", df)
```

```python
# 3. Combining DataFrames
df2 = pd.DataFrame({
    'ID': [4, 5],
    'Name': ['Rani', 'Kiran'],
    'Marks': [92, 81]
})
combined = pd.concat([df1, df2], ignore_index=True)
print("\nCombined DataFrame:\n", combined)
```

```python
# 4. Merging DataFrames
details = pd.DataFrame({
    'ID': [1, 2, 3],
    'City': ['Vizag', 'Hyderabad', 'Chennai']
})
merged = pd.merge(df1, details, on='ID')
print("\nMerged DataFrame:\n", merged)
```

```python
# 5. Reshaping using Melt
melted = pd.melt(
    df1, id_vars=['ID', 'Name'],
    value_vars=['Marks'],
    var_name='Subject', value_name='Score'
)
print("\nReshaped DataFrame:\n", melted)
```

```python
# 6. Pivoting a DataFrame
data = pd.DataFrame({
    'Name': ['Asha', 'Asha', 'Ravi', 'Ravi'],
    'Subject': ['Python', 'Java', 'Python', 'Java'],
    'Marks': [85, 80, 90, 88]
})
pivoted = data.pivot(
    index='Name', columns='Subject', values='Marks'
)
print("\nPivoted DataFrame:\n", pivoted)
```
